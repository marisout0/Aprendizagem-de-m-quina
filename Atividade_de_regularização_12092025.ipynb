## Analisar quais coeficientes permanecem conforme lambda muda

### Subtarefa:
Identificar quais coeficientes permanecem não nulos conforme o parâmetro de regularização (alpha) muda.

**Justificativa**:
Iterar pelos coeficientes do Lasso armazenados para cada valor de alpha e imprimir os índices dos coeficientes não nulos. Isso mostrará como o conjunto de coeficientes não nulos muda conforme a força de regularização aumenta.

```python
from sklearn.linear_model import Lasso

# Criar uma lista de valores de parâmetro de regularização
alphas = np.logspace(-4, 0, 100)  # Intervalo de 10^-4 a 10^0

# Inicializar uma lista para armazenar os coeficientes Lasso ajustados
lasso_coefficients = []

# Iterar pela lista de alphas, ajustar o Lasso e armazenar os coeficientes
for alpha in alphas:
    lasso = Lasso(alpha=alpha)
    lasso.fit(X, y)
    lasso_coefficients.append(lasso.coef_)

# Converter a lista de coeficientes para um array numpy para facilitar a análise posterior
lasso_coefficients = np.array(lasso_coefficients)

display(lasso_coefficients[:5]) # Exibir os primeiros 5 conjuntos de coeficientes como amostra
display(lasso_coefficients.shape) # Exibir a forma dos coeficientes armazenados
```

## Find optimal regularization parameter

### Subtask:
Identify the regularization parameter value that results in exactly 3 non-zero coefficients.


**Reasoning**:
Iterate through the lasso coefficients and find the first alpha that results in exactly 3 non-zero coefficients.



```python
# Definir uma pequena tolerância para considerar valores próximos de zero como zero
epsilon = 1e-6

# Inicializar optimal_alpha como None
optimal_alpha = None

# Iterar pelo array lasso_coefficients com seus valores alpha correspondentes
for i, coefficients in enumerate(lasso_coefficients):
    # Contar o número de coeficientes não nulos usando a tolerância definida
    non_zero_count = np.sum(np.abs(coefficients) > epsilon)

    # Verificar se o número de coeficientes não nulos é exatamente 3
    if non_zero_count == 3:
        # Se for, armazenar o valor de alpha correspondente e sair do loop
        optimal_alpha = alphas[i]
        break

# Exibir o optimal_alpha encontrado
print(f"O valor de alpha ótimo é: {optimal_alpha}")
```

Iterate through the Lasso coefficients for each alpha, identify which coefficients are zero, and compare their original values to the non-zero coefficients.



```python
# 1. Criar uma lista para armazenar quais coeficientes foram definidos como zero para cada parâmetro de regularização
zeroed_coefficients_mask = (np.abs(lasso_coefficients) < epsilon)

# 2. Analisar a relação entre a magnitude original e se eles foram zerados
# Podemos olhar para os coeficientes no alpha ótimo encontrado na etapa anterior como um exemplo representativo
optimal_alpha_index = np.where(alphas == optimal_alpha)[0][0]
coefficients_at_optimal_alpha = lasso_coefficients[optimal_alpha_index]
zeroed_at_optimal = (np.abs(coefficients_at_optimal_alpha) < epsilon)
non_zero_at_optimal = ~zeroed_at_optimal

print("Coeficientes verdadeiros:", true_coefficients)
print("Coeficientes Lasso no alpha ótimo:", coefficients_at_optimal_alpha)
print("Valores originais dos coeficientes zerados no alpha ótimo:", true_coefficients[zeroed_at_optimal])
print("Valores originais dos coeficientes não nulos no alpha ótimo:", true_coefficients[non_zero_at_optimal])

# 3. Determinar se os coeficientes zerados tenderam a ter valores originais menores ou maiores
# Comparar a média ou mediana dos valores absolutos dos coeficientes originais
mean_abs_true_zeroed = np.mean(np.abs(true_coefficients[zeroed_at_optimal]))
mean_abs_true_non_zero = np.mean(np.abs(true_coefficients[non_zero_at_optimal]))

print(f"Valor absoluto médio original dos coeficientes zerados no alpha ótimo: {mean_abs_true_zeroed}")
print(f"Valor absoluto médio original dos coeficientes não zerados no alpha ótimo: {mean_abs_true_non_zero}")
```

## Apply lasso with varying regularization

### Subtask:
Apply Lasso regression with a range of regularization parameter values to this new dataset.


**Reasoning**:
Apply Lasso regression with a range of regularization parameter values by iterating through different alpha values, fitting the model, and storing the coefficients.



```python
from sklearn.linear_model import Lasso

# Criar uma lista de valores de parâmetro de regularização
alphas = np.logspace(-4, 0, 100)  # Intervalo de 10^-4 a 10^0

# Inicializar uma lista para armazenar os coeficientes Lasso ajustados
lasso_coefficients = []

# Iterar pela lista de alphas, ajustar o Lasso e armazenar os coeficientes
for alpha in alphas:
    lasso = Lasso(alpha=alpha)
    lasso.fit(X, y)
    lasso_coefficients.append(lasso.coef_)

# Converter a lista de coeficientes para um array numpy para facilitar a análise posterior
lasso_coefficients = np.array(lasso_coefficients)

display(lasso_coefficients.shape) # Exibir a forma dos coeficientes armazenados
```

**Reasoning**:
The Lasso fitting process produced ConvergenceWarnings, which can sometimes indicate issues with the model fitting. To ensure reliable results, it's prudent to increase the maximum number of iterations for the Lasso model.



```python
# Inicializar uma lista para armazenar os coeficientes Lasso ajustados
lasso_coefficients = []

# Iterar pela lista de alphas, ajustar o Lasso e armazenar os coeficientes
for alpha in alphas:
    lasso = Lasso(alpha=alpha, max_iter=10000) # Increased max_iter
    lasso.fit(X, y)
    lasso_coefficients.append(lasso.coef_)

# Converter a lista de coeficientes para um array numpy para facilitar a análise posterior
lasso_coefficients = np.array(lasso_coefficients)

display(lasso_coefficients.shape) # Exibir a forma dos coeficientes armazenados
```

## Analyze lasso coefficients

### Subtask:
Analyze the obtained coefficients to see how many are set to zero and how the regularization affects them in this high-dimensional setting.


**Reasoning**:
Iterate through the lasso coefficients to count non-zero coefficients for each alpha, find the alpha that results in exactly 3 non-zero coefficients, identify the zeroed and non-zero original coefficients at this alpha, and compare their mean absolute values.



```python
# 1. Definir uma pequena tolerância para considerar valores próximos de zero como zero
epsilon = 1e-6

# Inicializar uma lista para armazenar as contagens de coeficientes não nulos para cada alpha
non_zero_counts = []

# 2. Iterar pelo array lasso_coefficients
for coefficients in lasso_coefficients:
    # Contar o número de coeficientes não nulos usando a tolerância definida
    non_zero_count = np.sum(np.abs(coefficients) > epsilon)
    non_zero_counts.append(non_zero_count)

# 3. Converter a lista de contagens para um array numpy
non_zero_counts = np.array(non_zero_counts)

# 4. Encontrar o valor de alpha que resulta em exatamente 3 coeficientes não nulos
# Inicializar optimal_alpha como None
optimal_alpha = None
optimal_alpha_index = None

# Iterar pelas contagens de não nulos com seus valores alpha correspondentes
for i, count in enumerate(non_zero_counts):
    # Verificar se o número de coeficientes não nulos é exatamente 3
    if count == 3:
        # Se for, armazenar o valor de alpha correspondente e seu índice e sair do loop
        optimal_alpha = alphas[i]
        optimal_alpha_index = i
        break

# Exibir o optimal_alpha encontrado
print(f"O valor de alpha ótimo que resulta em exatamente 3 coeficientes não nulos é: {optimal_alpha}")

# 5. Para o alpha ótimo, identificar os coeficientes zerados e não nulos originais
if optimal_alpha is not None:
    coefficients_at_optimal_alpha = lasso_coefficients[optimal_alpha_index]
    zeroed_at_optimal = (np.abs(coefficients_at_optimal_alpha) < epsilon)
    non_zero_at_optimal = ~zeroed_at_optimal

    print("\nValores originais dos coeficientes zerados no alpha ótimo:", true_coefficients[zeroed_at_optimal])
    print("Valores originais dos coeficientes não nulos no alpha ótimo:", true_coefficients[non_zero_at_optimal])

    # 6. Comparar os valores absolutos médios dos coeficientes originais zerados e não nulos
    if np.sum(zeroed_at_optimal) > 0:
        mean_abs_true_zeroed = np.mean(np.abs(true_coefficients[zeroed_at_optimal]))
        print(f"\nValor absoluto médio original dos coeficientes zerados no alpha ótimo: {mean_abs_true_zeroed:.4f}")
    else:
        mean_abs_true_zeroed = 0
        print("\nNão há coeficientes zerados no alpha ótimo.")


    if np.sum(non_zero_at_optimal) > 0:
        mean_abs_true_non_zero = np.mean(np.abs(true_coefficients[non_zero_at_optimal]))
        print(f"Valor absoluto médio original dos coeficientes não zerados no alpha ótimo: {mean_abs_true_non_zero:.4f}")
    else:
         mean_abs_true_non_zero = 0
         print("Não há coeficientes não zerados no alpha ótimo.")


    # 7. Determinar se os coeficientes zerados tenderam a ter valores originais menores ou maiores
    print("\nAnálise dos coeficientes zerados no alpha ótimo:")
    if mean_abs_true_zeroed < mean_abs_true_non_zero:
        print("Os coeficientes que foram definidos como zero pelo Lasso tenderam a ter valores absolutos originais menores em comparação com aqueles que permaneceram não nulos.")
    elif mean_abs_true_zeroed > mean_abs_true_non_zero:
         print("Os coeficientes que foram definidos como zero pelo Lasso tenderam a ter valores absolutos originais maiores em comparação com aqueles que permaneceram não nulos.")
    elif mean_abs_true_zeroed == mean_abs_true_non_zero and np.sum(zeroed_at_optimal) > 0:
        print("Os valores absolutos médios originais dos coeficientes zerados e não zerados são aproximadamente iguais.")
    else:
        print("Não foi possível comparar os coeficientes zerados e não zerados, pois não há ambos os tipos.")

else:
    print("\nNão foi encontrado um valor de alpha que resultasse em exatamente 3 coeficientes não nulos.")
```
