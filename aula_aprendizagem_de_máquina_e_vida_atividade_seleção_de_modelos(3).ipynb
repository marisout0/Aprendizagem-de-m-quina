# Task
Generate 100 data points based on the equation $y = A \times x^{5/3} + \text{erro}$, where A is a constant and erro is a random error term. Fit a linear regression model (polynomial of order one) to the generated data and visualize the results.

## Generate data

### Subtask:
Generate 100 data points for `x` and calculate `y` using the equation $y = A \times x^{5/3} + \text{erro}$.


**Reasoning**:
Generate 100 data points for x and calculate y based on the provided equation, including a random error term.



```python
import numpy as np

np.random.seed(42)
x = 10 * np.random.rand(100)
A = 5
erro = np.random.randn(100)
y = A * (x**(5/3)) + erroA×x5/3+erro
```

## Fit linear regression

### Subtask:
Fit a linear regression model (polynomial of order one) to the generated data.


**Reasoning**:
Import `LinearRegression`, reshape `x`, instantiate and fit the model.



```python
from sklearn.linear_model import LinearRegression

x_reshaped = x.reshape(-1, 1)
model = LinearRegression()
model.fit(x_reshaped, y)
```

## Visualize results

### Subtask:
Plot the generated data and the fitted linear regression line.


**Reasoning**:
Plot the generated data as a scatter plot and the fitted linear regression line on the same plot with appropriate labels and title.



```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(x, y, label='Generated Data')
plt.plot(x, model.predict(x_reshaped), color='red', label='Linear Regression Fit')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Generaty=A×x5/3+erroed Data and Linear Regression Fit')
plt.legend()
plt.show()
```

```python
print(f"Coefficient (slope): {model.coef_[0]}")
print(f"Intercept: {model.intercept_}")
```

## Initialize storage

### Subtask:
Create empty lists to store the coefficients 'a' (slope) and 'b' (intercept) from each iteration.


**Reasoning**:
Initialize empty lists to store the slopes and intercepts from each linear regression fit.



```python
slopes = []
intercepts = []
```

## Loop 100 times

### Subtask:
Generate 100 sets of data, fit a linear regression model to each set, and store the coefficients.


**Reasoning**:
Generate 100 sets of data, fit a linear regression model to each set, and store the coefficients in the previously initialized lists.



```python
for _ in range(100):
    x = 10 * np.random.rand(100)
    erro = np.random.randn(100)
    y = A * (x**(5/3)) + erro
A×x5/3+erro
    x_reshaped = x.reshape(-1, 1)
    model = LinearRegression()
    model.fit(x_reshaped, y)

    slopes.append(model.coef_[0])
    intercepts.append(model.intercept_)
```

## Fit quadratic model 100 times

### Subtask:
Generate 100 sets of data, fit a quadratic regression model to each set, and store the coefficients.


**Reasoning**:
Import PolynomialFeatures, create empty lists for quadratic coefficients, and loop 100 times to generate data, fit a quadratic model, and store the coefficients.



```python
from sklearn.preprocessing import PolynomialFeatures

quadratic_intercepts = []
quadratic_coefs = []

for _ in range(100):
    x = 10 * np.random.rand(100)
    erro = np.random.randn(100)
    y = A * (x**(5/3)) + erro

    x_reshaped = x.reshape(-1, 1)

    poly_features = PolynomialFeatures(degree=2)
    x_poly = poly_features.fit_transform(x_reshaped)

    model_quadratic = LinearRegression()
    model_quadratic.fit(x_poly, y)

    quadratic_intercepts.append(model_quadratic.intercept_)
    quadratic_coefs.extend(model_quadratic.coef_[1:])
```

## Analyze quadratic model coefficients

### Subtask:
Calculate and display the mean and variance of the stored quadratic model coefficients.


**Reasoning**:
Calculate and display the mean and variance of the stored quadratic model coefficients and intercepts.



```python
mean_quadratic_intercepts = np.mean(quadratic_intercepts)
variance_quadratic_intercepts = np.var(quadratic_intercepts)
mean_quadratic_coefs = np.mean(quadratic_coefs)
variance_quadratic_coefs = np.var(quadratic_coefs)

print(f"Mean of Quadratic Intercepts: {mean_quadratic_intercepts}")
print(f"Variance of Quadratic Intercepts: {variance_quadratic_intercepts}")
print(f"Mean of Quadratic Coefficients: {mean_quadratic_coefs}")
print(f"Variance of Quadratic Coefficients: {variance_quadratic_coefs}")
```
