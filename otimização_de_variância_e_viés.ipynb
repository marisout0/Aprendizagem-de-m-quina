## Explicação Didática: Random Forest e Gradient Boosting

Como vimos nos resultados, os modelos **Random Forest** e **Gradient Boosting** apresentaram um bom desempenho, equilibrando bem o viés e a variância. Mas o que são eles?

Ambos são exemplos de **métodos de *ensemble***. Pense em um *ensemble* como uma "equipe" de modelos menores trabalhando juntos para tomar uma decisão final, em vez de um único modelo tomando a decisão sozinho. A ideia é que a sabedagem da multidão (ou a combinação de vários modelos) pode ser mais precisa e robusta do que a de um único especialista (modelo).

### Random Forest (Floresta Aleatória)

Imagine que você quer prever o preço de uma casa. Em vez de pedir a opinião de apenas um corretor de imóveis (um único modelo de árvore de decisão), você pede a opinião de vários corretores (várias árvores de decisão).

*   **"Random" (Aleatória):** Cada corretor (árvore de decisão) não olha para *todas* as características da casa (número de quartos, tamanho do terreno, localização, etc.) e não olha para *todas* as casas disponíveis. Eles recebem um subconjunto aleatório das características e um subconjunto aleatório das casas para analisar. Isso garante que cada corretor (árvore) tenha uma perspectiva ligeiramente diferente.
*   **"Forest" (Floresta):** O conjunto de todas essas árvores de decisão individuais forma a "floresta aleatória".
*   **Como a decisão é tomada:** Para prever o preço de uma nova casa, você leva a casa para cada corretor (árvore) na sua floresta. Cada corretor dá seu palpite sobre o preço. No final, você junta todos os palpites e calcula a média (para problemas de regressão, como prever preços) ou escolhe o palpite que a maioria deu (para problemas de classificação).

**Benefício:** Ao combinar várias árvores de decisão treinadas em subconjuntos aleatórios de dados e características, a Random Forest reduz a **variância**. Se uma única árvore cometer um erro devido a um dado atípico, as outras árvores provavelmente não cometerão o mesmo erro, e a média dos palpites será mais próxima do valor correto.

### Gradient Boosting

Imagine que você tem um aluno (um modelo) que está tentando aprender a resolver problemas de matemática. Em vez de apenas dar a ele um monte de problemas e esperar que ele aprenda tudo de uma vez, você dá a ele um problema, vê onde ele errou, e então ensina outro "aluno" (outro modelo) a focar *apenas* nos erros que o primeiro aluno cometeu.

*   **Aprendizagem Sequencial:** No Gradient Boosting, os modelos são treinados **sequencialmente**. Cada novo modelo foca nos erros (os "resíduos") que o modelo anterior cometeu. É como ter uma equipe onde cada membro aprende com os erros do membro anterior para melhorar o desempenho geral.
*   **"Boosting":** A ideia é que modelos fracos (que sozinhos não são muito bons) podem ser "impulsionados" (boosted) para se tornarem um modelo forte quando combinados sequencialmente, focando nos pontos onde os modelos anteriores falharam.
*   **"Gradient":** O termo "Gradient" (gradiente) refere-se à técnica matemática usada para identificar e focar nos erros.

**Benefício:** O Gradient Boosting tende a reduzir tanto o **viés** quanto a **variância**, mas é particularmente eficaz na redução do viés, corrigindo iterativamente os erros dos modelos anteriores. No entanto, como cada modelo depende do anterior, ele pode ser mais sensível a dados ruídosos e propenso a overfitting se não for bem ajustado.

Em resumo, enquanto a Random Forest constrói várias árvores independentemente e combina seus resultados, o Gradient Boosting constrói árvores sequencialmente, onde cada nova árvore tenta corrigir os erros das árvores anteriores. Ambos são poderosas técnicas de ensemble que frequentemente alcançam alto desempenho em diversas tarefas de machine learning.

## Evaluate models for bias and variance

### Subtask:
Evaluate each model's performance on both the training and testing sets to assess bias and variance. Use metrics like Mean Squared Error (MSE), R-squared, or others appropriate for the task. Consider techniques like cross-validation to get more robust estimates of performance.


**Reasoning**:
Import the necessary evaluation metrics and initialize dictionaries for storing results. Then, iterate through the trained models, make predictions on both training and testing sets, calculate evaluation metrics, and store the results. Finally, print the results.



```python
from sklearn.metrics import mean_squared_error, r2_score

# Create empty dictionaries to store evaluation results
train_results = {}
test_results = {}

# Iterate through trained models and evaluate
for name, model in trained_models.items():
    # Make predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    # Calculate evaluation metrics (using MSE and R-squared as examples)
    train_mse = mean_squared_error(y_train, y_train_pred)
    test_mse = mean_squared_error(y_test, y_test_pred)
    train_r2 = r2_score(y_train, y_train_pred)
    test_r2 = r2_score(y_test, y_test_pred)

    # Store results
    train_results[name] = {'MSE': train_mse, 'R-squared': train_r2}
    test_results[name] = {'MSE': test_mse, 'R-squared': test_r2}

# Print evaluation results
print("\n--- Model Evaluation Results ---")
print("\nTraining Set Results:")
for name, metrics in train_results.items():
    print(f"{name}:")
    for metric, value in metrics.items():
        print(f"  {metric}: {value:.4f}")

print("\nTesting Set Results:")
for name, metrics in test_results.items():
    print(f"{name}:")
    for metric, value in metrics.items():
        print(f"  {metric}: {value:.4f}")
```

## Visualize results

### Subtask:
Visualize the performance metrics (e.g., MSE on training vs. testing sets) for each model to easily compare their bias-variance trade-off.


**Reasoning**:

```python
import matplotlib.pyplot as plt
import numpy as np

# Extract model names and MSEs
model_names = list(train_results.keys())
train_mses = [train_results[name]['MSE'] for name in model_names]
test_mses = [test_results[name]['MSE'] for name in model_names]
```

Create a bar chart to visualize the training and testing MSEs for each model to compare their performance and assess bias-variance trade-off.



```python
x = np.arange(len(model_names))  # the label locations
width = 0.35  # the width of the bars

fig, ax = plt.subplots(figsize=(12, 6))
rects1 = ax.bar(x - width/2, train_mses, width, label='Training MSE')
rects2 = ax.bar(x + width/2, test_mses, width, label='Testing MSE')

# Add some text for labels, title and custom x-axis tick labels, etc.
ax.set_xlabel('Model')
ax.set_ylabel('Mean Squared Error (MSE)')
ax.set_title('Training vs. Testing MSE for Different Models')
ax.set_xticks(x)
ax.set_xticklabels(model_names, rotation=45, ha="right")
ax.legend()

fig.tight_layout()
plt.show()
```
