```python
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
from mpl_toolkits.mplot3d import Axes3D

# Define the number of data points
n_samples = 100

# Generate two predictor variables (features)
X1 = 2 * np.random.rand(n_samples, 1)
X2 = 3 * np.random.rand(n_samples, 1)

# Define the coefficients and intercept for the linear model
beta1 = 5
beta2 = 2
intercept = 1

# Generate the dependent variable with some noise
y = intercept + beta1 * X1 + beta2 * X2 + np.random.randn(n_samples, 1)

# Display the first few points
print("Generated Data (first 5 points):")
for i in range(5):
    print(f"X1: {X1[i][0]:.2f}, X2: {X2[i][0]:.2f}, y: {y[i][0]:.2f}")

# Optional: Visualize the data (scatter plot of y vs X1 and y vs X2)
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.scatter(X1, y)
plt.xlabel("X1")
plt.ylabel("y")
plt.title("y vs X1")

plt.subplot(1, 2, 2)
plt.scatter(X2, y)
plt.xlabel("X2")
plt.ylabel("y")
plt.title("y vs X2")

plt.tight_layout()
plt.show()

# Add a constant to the predictor variables for the intercept
X = sm.add_constant(np.hstack((X1, X2)))

# Fit the linear regression model
model = sm.OLS(y, X)
results = model.fit()

# Print the summary of the regression results
print(results.summary())

# Visualize the regression plane in 3D
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Scatter plot of the data points
ax.scatter(X1, X2, y, c='blue', marker='o', label='Data Points')

# Create a meshgrid for the regression plane
x1_surf, x2_surf = np.meshgrid(np.linspace(X1.min(), X1.max(), 100),
                               np.linspace(X2.min(), X2.max(), 100))
# Calculate the corresponding y values on the regression plane
y_surf = results.params[0] + results.params[1] * x1_surf + results.params[2] * x2_surf

# Plot the regression plane
ax.plot_surface(x1_surf, x2_surf, y_surf, color='red', alpha=0.5, label='Regression Plane')

ax.set_xlabel('X1')
ax.set_ylabel('X2')
ax.set_zlabel('y')
ax.set_title('Multiple Linear Regression')
ax.legend()
plt.show()
```

## Define a function to run a single simulation

### Subtask:
Create a Python function that takes `n_samples` as input. Inside the function, generate the data (`X1`, `X2`, `y`) as done before, fit the `statsmodels` OLS model, calculate the predicted values (`y_pred`), and compute the Mean Squared Error (MSE) between the actual `y` and `y_pred`. The function should return the calculated MSE.


**Reasoning**:
Define a function `run_simulation` to encapsulate the data generation, model fitting, and MSE calculation process for a given number of samples.



```python
from sklearn.metrics import mean_squared_error

def run_simulation(n_samples):
    """
    Generates data, fits an OLS model, and calculates the Mean Squared Error.

    Args:
        n_samples: The number of data points to generate.

    Returns:
        The Mean Squared Error of the fitted model.
    """
    # Generate two predictor variables (features)
    X1 = 2 * np.random.rand(n_samples, 1)
    X2 = 3 * np.random.rand(n_samples, 1)

    # Define the coefficients and intercept for the linear model
    beta1 = 5
    beta2 = 2
    intercept = 1

    # Generate the dependent variable with some noise
    y = intercept + beta1 * X1 + beta2 * X2 + np.random.randn(n_samples, 1)

    # Add a constant to the predictor variables for the intercept
    X = sm.add_constant(np.hstack((X1, X2)))

    # Fit the linear regression model
    model = sm.OLS(y, X)
    results = model.fit()

    # Calculate predicted values
    y_pred = results.predict(X)

    # Calculate Mean Squared Error
    mse = mean_squared_error(y, y_pred)

    return mse
```

## Run the simulation for each sample size

### Subtask:
Use a loop to call the simulation function 100 times for `n_samples = 10`, 100 times for `n_samples = 100`, and 100 times for `n_samples = 1000`. Store the MSE results for each set of 100 runs in separate lists (e.g., `mse_10`, `mse_100`, `mse_1000`).


**Reasoning**:
Initialize lists to store MSEs and then run simulations for each sample size in separate loops.



```python
# Initialize lists to store MSE results
mse_10 = []
mse_100 = []
mse_1000 = []

# Run simulation 100 times for n_samples = 10
for _ in range(100):
    mse = run_simulation(10)
    mse_10.append(mse)

# Run simulation 100 times for n_samples = 100
for _ in range(100):
    mse = run_simulation(100)
    mse_100.append(mse)

# Run simulation 100 times for n_samples = 1000
for _ in range(100):
    mse = run_simulation(1000)
    mse_1000.append(mse)
```

## Define a function to run a single simulation

### Subtask:
Create a Python function that takes `n_samples` as input. Inside the function, generate the data (`X1`, `X2`, `y`) as done before, fit the `statsmodels` OLS model, calculate the predicted values (`y_pred`), and compute the Mean Squared Error (MSE) between the actual `y` and `y_pred`. The function should return the calculated MSE.

**Reasoning**:
Define a function `run_simulation` to encapsulate the data generation, model fitting, and MSE calculation process for a given number of samples.

```python
from sklearn.metrics import mean_squared_error
import numpy as np
import statsmodels.api as sm

def run_simulation(n_samples):
    """
    Generates data, fits an OLS model, and calculates the Mean Squared Error,
    estimated beta1, and its standard error.

    Args:
        n_samples: The number of data points to generate.

    Returns:
        A tuple containing:
            - The Mean Squared Error of the fitted model.
            - The estimated coefficient for beta1.
            - The standard error of the estimated beta1.
    """
    # Generate two predictor variables (features)
    X1 = 2 * np.random.rand(n_samples, 1)
    X2 = 3 * np.random.rand(n_samples, 1)

    # Define the coefficients and intercept for the linear model
    beta1 = 5
    beta2 = 2
    intercept = 1

    # Generate the dependent variable with some noise
    y = intercept + beta1 * X1 + beta2 * X2 + np.random.randn(n_samples, 1)

    # Add a constant to the predictor variables for the intercept
    X = sm.add_constant(np.hstack((X1, X2)))

    # Fit the linear regression model
    model = sm.OLS(y, X)
    results = model.fit()

    # Calculate predicted values
    y_pred = results.predict(X)

    # Calculate Mean Squared Error
    mse = mean_squared_error(y, y_pred)

    # Get the estimated beta1 coefficient and its standard error
    # The order of coefficients in statsmodels is const, X1, X2
    estimated_beta1 = results.params[1]
    std_err_beta1 = results.bse[1]


    return mse, estimated_beta1, std_err_beta1
```

### Subtask:
Use a loop to call the simulation function 100 times for `n_samples = 10`, 100 times for `n_samples = 100`, and 100 times for `n_samples = 1000`. Store the MSE results for each set of 100 runs in separate lists (e.g., `mse_10`, `mse_100`, `mse_1000`).

**Reasoning**:
Initialize lists to store MSEs and then run simulations for each sample size in separate loops.

```python
# Initialize lists to store results
mse_10 = []
beta1_10 = []
std_err_beta1_10 = []

mse_1000 = []
beta1_1000 = []
std_err_beta1_1000 = []

mse_100000 = [] # Changed to 100000 samples
beta1_100000 = [] # Changed to 100000 samples
std_err_beta1_100000 = [] # Changed to 100000 samples


# Run simulation 100 times for n_samples = 10
for _ in range(100):
    mse, estimated_beta1, std_err_beta1 = run_simulation(10)
    mse_10.append(mse)
    beta1_10.append(estimated_beta1)
    std_err_beta1_10.append(std_err_beta1)


# Run simulation 100 times for n_samples = 1000
for _ in range(100):
    mse, estimated_beta1, std_err_beta1 = run_simulation(1000)
    mse_1000.append(mse)
    beta1_1000.append(estimated_beta1)
    std_err_beta1_1000.append(std_err_beta1)


# Run simulation 100 times for n_samples = 100000
for _ in range(100):
    mse, estimated_beta1, std_err_beta1 = run_simulation(100000) # Changed to 100000 samples
    mse_100000.append(mse)
    beta1_100000.append(estimated_beta1)
    std_err_beta1_100000.append(std_err_beta1)
```
